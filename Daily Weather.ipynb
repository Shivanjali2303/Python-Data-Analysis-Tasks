```python
import pandas as pd
import numpy as np
df = pd.read_csv("Daily Weather.csv")
print(df)
```

```python
# Handle Missing Values (Fill with median)
df['air_pressure_9am'] = df['air_temp_9am'].fillna(df['avg_wind_direction_9am'].median())
```

```python
#Drop Rows with missing values
df = df.dropna()
```

```python
df = df.dropna(subset=['avg_wind_direction_9am'])
```

```python
# Normalize a numerical column (MinMaxScaler)
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df['air_pressure_9am'] = scaler.fit_transform(df[['air_temp_9am']])
```

```python
# Standarddize a numerical column (Standardscaler)
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df['rain_accumulation_9am'] = scaler.fit_transform(df[['rain_duration_9am']])
```

```python
#Create a new column using mathematical operation
df['new_column'] = df['rain_accumulation_9am'] ** 2
```

```python
# Change data type (Float Int)

df['rain_duration_9am'] = df['rain_duration_9am'].fillna(0)
df['rain_accumulation_9am'] = df['rain_duration_9am'].astype(int)

```

```python
# Detect and remove outliers using IQR method
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Daily Weather.csv")
# Example: Detect outliers in 'age' column

Q1 = df['avg_wind_speed_9am'].quantile(0.25)
Q3 = df['avg_wind_direction_9am'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter dataset
df_no_outliers = df[(df['avg_wind_direction_9am'] >= lower_bound) & (df['avg_wind_speed_9am'] <= upper_bound)]

print("Original Shape:", df.shape)
print("After Removing Outliers:", df_no_outliers.shape)

```

```python
#19Create a Stacked Bar Chart Comparing Two Categorical Variable
import pandas as pd
df = pd.read_csv("Daily Weather.csv")
ct = pd.crosstab(df['air_pressure_9am'], df['air_temp_9am'])

ct.plot(kind='bar', stacked=True, figsize=(8,5))
plt.title("air_pressure_9am  vs air_temp_9am")
plt.xlabel("air_pressure_9am")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```
