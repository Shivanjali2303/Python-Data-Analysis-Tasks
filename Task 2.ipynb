```python
# a) If-Else Condations

# 1) Grade code

p = 50

if p > 75:
    print("Grade: O")
elif 60 < p <= 75:
    print("Grade: A")
elif 50 < p <= 60:
    print("Grade: B")
elif 35 < p <= 50:
    print("Grade: C")
else:
    print("Fail")
```

```python
# 2) Divisibility by 2 and/or 3
n = 7

if n % 2 == 0 and n % 3 == 0:
    print("Divisible by 2 and 3")
elif n % 2 == 0:
    print("Divisible by 2, not by 3")
elif n % 3 == 0:
    print("Divisible by 3, not by 2")
else:
    print("Not divisible by 2 and 3")
```

```python
# Additional If / If-Else / If-Elif-Else Examples

# Example 1: Check positive or negative
num = -5
if num > 0:
    print("Positive")
else:
    print("Negative or Zero")

# Example 2: Voting eligibility
age = 18
if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible")

# Example 3: Even or Odd
n = 9
if n % 2 == 0:
    print("Even")
else:
    print("Odd")

# Example 4: Largest of two numbers
a, b = 10, 20
if a > b:
    print("a is greater")
else:
    print("b is greater")

# Example 5: Temperature check
temp = 30
if temp > 35:
    print("Hot")
elif temp > 20:
    print("Warm")
else:
    print("Cold")
```

```python
#b) For Loop

#1) Print odd numbers between 20 and 80 without if
for i in range(21, 80, 2):
    print(i)
```

```python
# 2) Create list from 1 to 20
lst = []
for i in range(1, 21):
    lst.append(i)
print(lst)
```

```python
# 3) Create list from 20 to 1 without reverse
lst = []
for i in range(20, 0, -1):
    lst.append(i)
print(lst)
```

```python
# 4) Cube of odd values between 20 to 40
for i in range(21, 40, 2):
    print(i, "cube =", i**3)
```

```python
# 5) Friends name and age
names = ['a', 'b', 'c', 'd', 'e']
ages = [20, 21, 23, 25, 24]

for i in range(len(names)):
    print(f"my name is {names[i]}, my age is {ages[i]}")
```

Additional For Loop Examaples


```python
#Additional For Loop Examaples

# Example 1: Square of numbers
for i in range(1, 6):
    print(i*i)

# Example 2: Sum of numbers
total = 0
for i in range(1, 11):
    total += i
print(total)

# Example 3: Print characters of string
for ch in "Python":
    print(ch)

# Example 4: Multiplication table of 5
for i in range(1, 11):
    print(5*i)

# Example 5: Count vowels
count = 0
for ch in "education":
    if ch in "aeiou":
        count += 1
print(count)
```

```python
# c) If + For Loop + Data Types

# Swap first and last element of list
lst = [1, 2, 3, 4, 5]
lst[0], lst[-1] = lst[-1], lst[0]
print(lst)
```

```python
# Count length of string
s = "Python"
count = 0
for _ in s:
    count += 1
print("Length:", count)
```

```python
# Sum of non-negative integers
lst = [1, 4, -5, -20, 10]
total = 0
for i in lst:
    if i >= 0:
        total += i
print("Sum:", total)
```

```python
# Factorial
n = 3
fact = 1
for i in range(1, n+1):
    fact *= i
print("Factorial:", fact)
```

```python
# d) While Loop

# 1) Odd-Even using while loop
n = 1
while n <= 10:
    if n % 2 == 0:
        print(n, "Even")
    else:
        print(n, "Odd")
    n += 1
```

```python

# 2) Star pattern using while
i = 1
while i <= 5:
    print("* " * i)
    i += 1
```

```python
# 3) List 1 to 20 using while
lst = []
i = 1
while i <= 20:
    lst.append(i)
    i += 1
print(lst)
```

```python
# 4) List 20 to 1 using while
lst = []
i = 20
while i >= 1:
    lst.append(i)
    i -= 1
print(lst)
```

```python
# 5) break, continue, pass example
i = 1
while i <= 10:
    if i == 3:
        i += 1
        continue
    if i == 8:
        break
    print(i)
    i += 1
```

Additional While Loop Examples

```python
# Example 1: Sum until 5
i = 1
total = 0
while i <= 5:
    total += i
    i += 1
print(total)

# Example 2: Countdown
i = 5
while i > 0:
    print(i)
    i -= 1

# Example 3: Print even numbers
i = 2
while i <= 20:
    print(i)
    i += 2

# Example 4: String traversal
s = "AI"
i = 0
while i < len(s):
    print(s[i])
    i += 1

# Example 5: pass statement
i = 1
while i <= 5:
    if i == 3:
        pass
    print(i)
    i += 1
```
