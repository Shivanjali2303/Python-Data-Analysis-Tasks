```python
#1. Numbers in python


# Integer, Float,and Complex numbers
a = 10       #Integer
b = 5.5     # Float
c = 2 + 3j  #Complex

# Arithmetic operations
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Complex Number:", c)
```

```python
#2. Operators in python 

x = 10
y = 5

# Arithmetic operators
print(x + y)
print(x - y)

# Comparison operators
print(x == y)
print(x != y)
print(x > y)

# Logical operators
print(x > 5 and y < 10)
print(x > 20 or y < 10)
```

```python
#3.Lists in python

my_list = [1, 2, 3, 4, 5]

# Accessing elements
print(my_list[0])
print(my_list[1:4])

# Modifying elements
my_list[0] = 10

# Adding elements
my_list.append(6)
my_list.insert(1, 20)

# Removing elements
my_list.remove(3)
my_list.pop()
# my_list.clear()
# Other methods
my_list.sort()
my_list.reverse()
print(my_list.count(2))
print(my_list.index(2))
copy_list = my_list.copy()

print(my_list)
print(copy_list)
```

```python
#4. Tuples in python
my_tuple = (1, 2, 3, 4, 2)

# Accessing elements
print(my_tuple[0])
print(my_tuple[1:4])

# Concatenation
new_tuple = my_tuple + (5, 6)

# Repetition
repeat_tuple = my_tuple * 2

# Count and index
print(my_tuple.count(2))
print(my_tuple.index(3))

print(new_tuple)
print(repeat_tuple)
```

```python
#5. Sets in python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
# Adding and removing elements
set1.add(4)
set1.remove(2)

#set operations
print(set1.union(set2))
print(set1.intersection(set2))
print(set1.difference(set2))
```

```python
#6.Dictionaries in python


my_dict = {'name': 'Python', 'version': 3}

# Accessing values
print(my_dict.get('name'))

# Dictionary methods
print(my_dict.keys())
print(my_dict.values())
print(my_dict.items())

# Updating and removing
my_dict.update({'year': 2025})
my_dict.pop('version')

print(my_dict)
```

```python
#7. Common python errors

# IndentationError example (fixed)
if True:
    print("Correct indentation")

# NameError example (fixed)
a = 10
print(a)

# ValueError example (handled)
try:
    int("abc")
except ValueError:
    print("ValueError handled")

# TypeError example (handled)
try:
    print(5 + "a")
except TypeError:
    print("TypeError handled")

# IndexError example (handled)
lst = [1, 2, 3]
try:
    print(lst[5])
except IndexError:
    print("IndexError handled")

# KeyError example (handled)
d = {'a': 1}
try:
    print(d['b'])
except KeyError:
    print("KeyError handled")
```
