```python
#Merging and joining
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Amit', 'Neha', 'Rahul']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 88]
})

merged_df = pd.merge(df1, df2, on='ID')
print(merged_df)
```

```python
#Concatenation
pd.concat([df1, df2], axis=1)
```
