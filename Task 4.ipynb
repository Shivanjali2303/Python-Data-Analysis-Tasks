```python
#2) arrays in numpy
import numpy as np
#using array 
a = np.array([1, 2, 3, 4])
#Zeros 
b = np.zeros(5)
# ones
c = np.ones(3)
#arange
d = np.arange(1, 10, 2)
#Linspace
e = np.linspace(0, 1, 5)
print(a, b, c, d, e)
```

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[0])    
print(arr[1:4])   # Slicing
```

```python
#3 Array Operations
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print(x + y)
print(x - y)
print(x * y)
print(x / y)
```

```python
arr = np.array([1, 2, 3])
print(arr + 10)
```

```python
print(np.sqrt(arr))
print(np.sin(arr))
print(np.exp(arr))
```

```python
#4 Array shapes and 

# shape and dimensions
arr = np.array([[1, 2, 3], [4, 5, 6]])

print(arr.shape)
print(arr.ndim)
```

```python
#Reshape,Flatten, Transpose
print(arr.reshape(3, 2))
print(arr.flatten())
print(arr.transpose())
```

```python
#Concatenation and splitting
a = np.array([1, 2])
b = np.array([3, 4])

print(np.concatenate((a, b)))
print(np.vstack((a, b)))
print(np.hstack((a, b)))

arr = np.array([1, 2, 3, 4])
print(np.split(arr, 2))
```

```python
# Array Manipulation
#multi-dimensional indexing
arr = np.array([[1, 2], [3, 4]])
print(arr[0, 1])
```

```python
# Masking (Boolean Indexing)
arr = np.array([10, 20, 30, 40])
print(arr[arr > 20])
```

```python
#Fancy Indexing
arr = np.array([10, 20, 30, 40])
print(arr[[0, 2]])
```

```python
# Mathematical Functions 
#universal functions (ufuncs)
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.add(a, b))
print(np.subtract(a, b))
print(np.multiply(a, b))
```

```python
#statistical function
data = np.array([10, 20, 30, 40, 50])

print(np.mean(data))
print(np.median(data))
print(np.std(data))
print(np.var(data))
print(np.min(data))
print(np.max(data))
```

```python
#Linear algebra
mat = np.array([[1, 2], [3, 4]])

print(np.dot(mat, mat))
print(np.linalg.inv(mat))
print(np.linalg.eig(mat))
```

```python
#Random Number Generation
np.random.seed(10)

print(np.random.rand(3))
print(np.random.randint(1, 10, 5))

arr = np.array([1, 2, 3, 4])
np.random.shuffle(arr)
print(arr)

print(np.random.permutation(arr))
```

```python
#Broadcasting
a = np.array([[1, 2, 3],
              [4, 5, 6]])

b = np.array([10, 20, 30])

print(a + b)
```
