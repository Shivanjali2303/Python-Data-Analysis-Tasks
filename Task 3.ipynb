```python
 #1)Function to find the maximum number from given numbers
def find_max(numbers):
    return max(numbers)

# Example
print(find_max([10, 25, 7, 99, 45]))  
```

```python
#2) Function to sum all numbers in a list
def sum_list(numbers):
    return sum(numbers)

# Sample List
sample_list = [8, 2, 3, 0, 7]
print("Sum =", sum_list(sample_list))  
```

```python
#3) Function to return a list with distinct elements
def unique_list(lst):
    return list(set(lst))

# Sample List
sample = [1, 2, 3, 3, 3, 3, 4, 5]
print("Unique List =", unique_list(sample))
```

```python
#4) Function to find total number of combinations
from itertools import combinations

def total_combinations(elements):
    result = []
    for i in range(2, len(elements) + 1):
        result.extend(combinations(elements, i))
    return result

# Example
items = ['a', 'b', 'c']
comb = total_combinations(items)
print("Total combinations =", len(comb))
print(comb)
```

```python
#5) Function to find total number of permutations
from itertools import permutations

def total_permutations(elements):
    return list(permutations(elements))

# Example
perm = total_permutations(['a', 'b', 'c'])
print("Total permutations =", len(perm))
print(perm)
```

```python
#6) Function to count vowels and consonants

def to_uppercase(word):
    return word.upper()

# Example
print("Result is =", to_uppercase("statistics"))
```

```python
 #7) Count lowercase and uppercase letters
def count_upper_lower(text):
    upper = lower = 0
    for ch in text:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1
    return upper, lower

# Example
text = "STatiStiCS"
u, l = count_upper_lower(text)
print("Upper case :", u)
print("Lower case :", l)
```

```python
# ---------- Function Definitions ----------

def find_max(numbers):
    return max(numbers)

def sum_list(numbers):
    return sum(numbers)

def unique_list(lst):
    return list(set(lst))

def count_vowels_consonants(word):
    vowels = "aeiou"
    v = 0
    c = 0
    for ch in word.lower():
        if ch.isalpha():
            if ch in vowels:
                v += 1
            else:
                c += 1
    return v, c

def to_uppercase(word):
    return word.upper()

def count_upper_lower(text):
    upper = 0
    lower = 0
    for ch in text:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1
    return upper, lower
```

```python
# Example 1
print(find_max([3, 6, 1, 9]))  

# Example 2
print(sum_list([1, 2, 3, 4, 5]))

# Example 3
print(unique_list([10, 10, 20, 30, 30]))

# Example 4
print(count_vowels_consonants("education"))

# Example 5
print(to_uppercase("python"))
```
